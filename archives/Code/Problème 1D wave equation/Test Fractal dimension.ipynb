# Test — Dimension fractale des outputs

On applique la méthode **two-NN** (Facco et al.) au nuage 3D formé par les outputs
`(u_dot, u_xx, delta_u)` du dataset généré par le notebook 01.

**Objectif** : estimer la dimension intrinsèque (en essayant de calculer la diension fractale via la méthode two_NN) de l'espace de sortie. Si elle est nettement
inférieure à 3, ça veut dire que les trois outputs sont liés (sous-variété),
et donc que le réseau a moins d'information à apprendre que ce qu'on pourrait croire.

---
## 1. Imports et fonction `two_nn`

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import time
from sklearn.neighbors import NearestNeighbors

np.random.seed(42)

def two_nn(data):
    nn = NearestNeighbors(n_neighbors=3)   # 3 car le point lui-même compte
    nn.fit(data)
    
    distances, _ = nn.kneighbors(data)
    distances = distances[:, 1:]            # on enlève la distance à soi-même
    
    r1 = distances[:, 0]
    r2 = distances[:, 1]
    
    valid = r1 > 0
    
    mu = r2[valid] / r1[valid]
    
    mu_sorted = np.sort(mu)
    
    N = len(mu_sorted)
    
    F_emp = np.arange(1, N + 1) / N
    
    mu_sorted = mu_sorted[:-1]
    F_emp     = F_emp[:-1]
    
    x = np.log(mu_sorted)
    y = -np.log(1 - F_emp)
    
    d = np.sum(x * y) / np.sum(x**2)
    
    return d
```

---
## 2. Chargement des données

```python
# Chargement du dataset complet généré par le notebook 01
df = pd.read_csv("wave_beam_dataset.csv")

# Échantillonnage : 1.5M points c'est trop pour two-NN
# 30k est largement suffisant pour une estimation stable
N_SAMPLE = 30000        # Maybe to modify to change the accuracy of the computation of the dimension with two_NN
sample = df.sample(N_SAMPLE, random_state=42)   #We selectionate N_SAMPLE number of points randomly 

OUTPUT_COLS_N = ["u_dot_n", "u_xx_n"]
data_outputs = sample[OUTPUT_COLS_N].values
d = two_nn(data_outputs)
print(f"Dimension fractale estimée pour {OUTPUT_COLS_N} : {d:.3f}")

OUTPUT_COLS_N = ["u_dot_n", "u_xx_n", "delta_u_n"]
data_outputs = sample[OUTPUT_COLS_N].values
d = two_nn(data_outputs)
print(f"Dimension fractale estimée pour {OUTPUT_COLS_N} : {d:.3f}")

OUTPUT_COLS_N = ["u_n","u_dot_n", "u_xx_n", "delta_u_n"]
data_outputs = sample[OUTPUT_COLS_N].values
d = two_nn(data_outputs)
print(f"Dimension fractale estimée pour {OUTPUT_COLS_N} : {d:.3f}")

OUTPUT_COLS_N = ["u","u_dot", "u_xx"]
data_outputs = sample[OUTPUT_COLS_N].values
d = two_nn(data_outputs)
print(f"Dimension fractale estimée pour {OUTPUT_COLS_N} : {d:.3f}")

OUTPUT_COLS_N = ["u"]
data_outputs = sample[OUTPUT_COLS_N].values
d = two_nn(data_outputs)
print(f"Dimension fractale estimée pour {OUTPUT_COLS_N} : {d:.3f}")

OUTPUT_COLS_N = ["u_dot"]
data_outputs = sample[OUTPUT_COLS_N].values
d = two_nn(data_outputs)
print(f"Dimension fractale estimée pour {OUTPUT_COLS_N} : {d:.3f}")

OUTPUT_COLS_N = ["u_xx"]
data_outputs = sample[OUTPUT_COLS_N].values
d = two_nn(data_outputs)
print(f"Dimension fractale estimée pour {OUTPUT_COLS_N} : {d:.3f}")

OUTPUT_COLS_N = ["u","u_dot"]
data_outputs = sample[OUTPUT_COLS_N].values
d = two_nn(data_outputs)
print(f"Dimension fractale estimée pour {OUTPUT_COLS_N} : {d:.3f}")

OUTPUT_COLS_N = ["u","u_xx"]
data_outputs = sample[OUTPUT_COLS_N].values
d = two_nn(data_outputs)
print(f"Dimension fractale estimée pour {OUTPUT_COLS_N} : {d:.3f}")
```
